La mediana obtenida fue de 12, que es significativamente inferior  a la media, podemos ver que la distribución de los datos esta sesgada a la derecha por comentarios muy largos. <span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 1.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">[Característica adicional 1: Proporción de tokens en mayúsculas]</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
counts_upper = []
for comentario in doc:
    # Contamos cuántos tokens son palabras (alfabéticos)
    solo_palabras = [t for t in comentario if t.is_alpha]
    total_palabras = len(solo_palabras)
    
    if total_palabras > 0:
        # De esas palabras, ¿cuántas están en mayúsculas?
        mayus = sum(1 for t in solo_palabras if t.is_upper)
        counts_upper.append(mayus / total_palabras)
    else:
        counts_upper.append(0.0)
upper_odio = [c for c, v in zip(counts_upper, value) if v > 0]
upper_no_odio = [c for c, v in zip(counts_upper, value) if v == 0]


print(f"Odio - Promedio: {np.mean(upper_odio):.4f}, Mediana: {np.median(upper_odio)}, Desviación: {np.std(upper_odio):.4f}")
print(f"No Odio - Promedio: {np.mean(upper_no_odio):.4f}, Mediana: {np.median(upper_no_odio)}, Desviación: {np.std(upper_no_odio):.4f}")
```

<b>Incluye aquí, debajo de la línea, la explicación de la característica propuesta y su motivación. Incluye también una explicación del código fuente aportado.</b>
<hr><b></b>
En esta etapa se analizó la hipótesis de que el uso de palabras completamente en mayúsculas podría ser un indicador de agresividad en los comentarios de odio. Para ello, se calculó la proporción de tokens que cumplen la condición token.is_upper sobre el total de palabras alfabéticas (token.is_alpha) de cada comentario, manejando los casos de comentarios sin palabras alfabéticas con un valor de 0.0 para evitar divisiones por cero.
Los resultados mostraron proporciones prácticamente nulas en ambos grupos: 0.0000 para odio y 0.0001 para no odio. Las medianas fueron 0.0 en ambos casos y las desviaciones estándar despreciables (0.0000 para odio y 0.0104 para no odio), lo que indica que el usuario típico no utiliza palabras completas en mayúsculas, independientemente de la intención del mensaje.

 

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 2.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Análisis de la proporción de signos de puntuación</span>

<hr><b>Característica adicional 2: Análisis de la proporción de signos de puntuación

Los resultados muestran que los comentarios sin odio presentan una proporción de puntuación ligeramente superior (0.16) frente a los de odio (0.14). Las medianas también reflejan esta tendencia (0.13 vs 0.12), con desviaciones estándar muy similares (0.13 y 0.12 respectivamente).
Esta pequeña diferencia sugiere que el discurso de odio en este corpus no se caracteriza necesariamente por un uso exagerado de signos de puntuación, sino más bien por una estructura levemente más simple o menos puntuada que el discurso neutro. Al igual que en las métricas anteriores, la cercanía entre estos valores indica que la densidad de puntuación por sí sola no es un predictor fuerte para identificar el odio.

```python
# Incluye aquí el código generado para poder responder a tu pregunta
from scipy.stats import mannwhitneyu
from scipy.stats import chi2_contingency
def mannwhitney_test(df, feature, target="label", alpha=0.05):
    grupo_odio = df[df[target] == 1][feature].dropna()
    grupo_no_odio = df[df[target] == 0][feature].dropna()
    
    stat, p_value = mannwhitneyu(
        grupo_odio,
        grupo_no_odio,
        alternative="two-sided"
    )
    
    n1 = len(grupo_odio)
    n2 = len(grupo_no_odio)
    
    mean_U = n1 * n2 / 2
    std_U = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
    z = (stat - mean_U) / std_U
    r = z / np.sqrt(n1 + n2)
    
    if abs(r) < 0.1:
        efecto = "pequeño (casi nulo)"
    elif abs(r) < 0.3:
        efecto = "pequeño"
    elif abs(r) < 0.5:
        efecto = "medio"
    else:
        efecto = "grande"
    
    print(f"\n=== Test Mann-Whitney para {feature} ===")
    print(f"U statistic: {stat:.4f}")
    print(f"p-value: {p_value:.6f}")
    print(f"Tamaño del efecto (r): {r:.4f} -> {efecto}")
    
    if p_value < alpha:
        print("Conclusión: Diferencia estadísticamente significativa (rechazo H0)")
    else:
        print("Conclusión: No significativa (no se rechaza H0)")

label = [1 if intensidad > 0 else 0 for intensidad in value]

df = pd.DataFrame({
    "prop_p2": counts_p2,
    "prop_adj": counts_adj,
    "tiene_ner": tiene_entidades,
    "label": label
})

df.index.name = "comentario"
```
