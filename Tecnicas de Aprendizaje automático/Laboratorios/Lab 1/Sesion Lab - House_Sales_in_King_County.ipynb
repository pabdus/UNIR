El dataset tiene **21613** instancias. Cada instancia se compone de **21** variables:
- id: Identificador único para cada casa vendida.
-date: Fecha en la que se realizó la venta de la propiedad.
- price: Precio de venta de la vivienda (variable objetivo).
- bedrooms: Número de habitaciones de la casa.
- bathrooms: Número de baños, donde .5 indica un baño sin ducha.
- sqft_living: Pies cuadrados de espacio habitable interior.
- sqft_lot: Pies cuadrados de la superficie total del terreno.
- floors: Número de pisos o niveles que tiene la casa.
- waterfront: Variable binaria que indica si la casa tiene vistas al mar/agua.
- view: Índice de 0 a 4 sobre qué tan buena es la vista desde la propiedad.
- condition: Índice de 1 a 5 que mide la condición estructural de la casa.
- grade: Índice de 1 a 13 que califica el nivel de construcción y diseño.
- sqft_above: Pies cuadrados de espacio interior que están por encima del nivel del suelo.
- sqft_basement: Pies cuadrados de espacio interior que están por debajo del nivel del suelo.
- yr_built: Año en el que se construyó originalmente la vivienda.
- yr_renovated: Año de la última remodelación (0 si nunca se ha renovado).
- zipcode: Código postal del área donde se ubica la casa.
- lat: Coordenada de latitud geográfica.
- long: Coordenada de longitud geográfica.
- sqft_living15: Pies cuadrados promedio del espacio habitable de los 15 vecinos más cercanos.
- sqft_lot15: Pies cuadrados promedio del terreno de los 15 vecinos más cercanos.
